# Elasticity  

## Definition  

A material is elastic if and only if the magnitude of the weight stretching it is proportional to the amount that it is being stretched.  Put more formally, a material is elastic if and only if:  

\begin{equation}\tag{1}
    F \propto x
\end{equation}  

where $F$ is the magnitude of said weight or force $\mathbf{F}$ and $x$ is the amount that our material is being stretched.

The constant of proportionality between $F$ and $x$ is known as the elastic constant, denoted as $\varepsilon$.  Via equation (1),  

\begin{equation}\tag{2}
    F = \varepsilon x
\end{equation}

## Elastic Potential  

If we were to use a rubber band that satisfies equation (1) as a slingshot to launch a marble -- or some other suitable object -- into the air the work done by the rubber band to accelerate our projectile is given by:  

\begin{equation}\tag{3}
    W = \int_{x_1}^{x_2} \varepsilon x\,\mathrm{d}x
\end{equation}

where $W$ -- A.K.A the *elastic potential* -- is the work done by the rubber band, $x_2$ is the amount that the band is streched just prior to launch, and $x_1$ is the amount that said band is streched after it is done continuously accelerating the projectile.

Of course, equation (3) works out to:  

$$ W = \varepsilon\,\frac{x_2^2 - x_1^2}{2}$$

## Examples

It should be noted that the [definition](#Definition) of elasticity found in the notebook delineates a foundational principle and is thus not subject to argument.  What is subject to empirical research is whether elastic materials exist.  The point of experiments is to identify such materials, which is what we will be doing to give some examples of elastic materials.

### Example 1  

Our first example is a rubber band found in my house!  To show that this object satisfies equation (1) we will be using a 
* cylindrical container,
* water,
* a ruler, and of course
* said rubber band.

The argument for this being elastic accurately assumes that the density of the water in the container is uniform.

First, to make things easier, we want to ensure that our container is light enough to have no significant impact on $x$.  We do this by tying the empty container to the rubber band and then hanging the band from a door knob or any other platform that allows the weight to pull the rubber band directly down.  If the rubber band doesn't stretch after this then $x=0$ under such circumstances and thus the weight of the empter container plus the object used to tie it to the band is negligible.

Next, we will add varying measured amounts of water to our container and measure the degree to which the band is stretched by the weight of this water.  Note that it is the volume of the water that will be measured;  The weight of said water is, via the aforementioned assumption about density, proportional to it's volume.

\begin{equation}\tag{1.1}
    \delta g = \frac{F}{V} \quad\therefore \quad F = \delta g V
\end{equation}

where $\delta$ and $g$ are the density of the water and acceleration due to gravity, which are both constant under the circumstances of this experiment.  Furthermore, via the fact that $\delta, g =\mathit{const}$ and the above proposition, $F\propto V$.

This proportionality, the one implied by the proposition of (1.1), in conjunction with proportionality (1), implies that $x$ is proportional to $V$ if and only if the rubber band is elastic.  

\begin{equation}\tag{1.2}
    V \propto x
\end{equation}

Therefore, it is the proportionality stated in (1.2) that we are trying to verify regarding our rubber band and the volume of the water whose weight is stretching it.

So now, our task is to make sure that the ratio between the volume of the water and the degree to which the rubber band is stretched remains approximately constant, which would suggest (1.2) and, in turn, that the rubber band is elastic.
